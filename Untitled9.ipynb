```python
# main.py
import tkinter as tk
from tkinter import ttk
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
from numba import njit, prange

# --- Numba 燒蝕核心 ---
@njit(parallel=True, fastmath=True)
def compute_single_pass_ablation(x_grid_um, y_grid_um, x_spots_um, y_spots_um, current_depth, F0, effective_F_th, delta_um, w_spot_um, zR_um, D_sat=12.0):
    nx, ny, n_spots = len(x_grid_um), len(y_grid_um), len(x_spots_um)
    updated_depth = current_depth.copy()
    w_sq = w_spot_um * w_spot_um
    cutoff_r_sq = 3.5 * 3.5 * w_sq

    for s in prange(n_spots):
        xs, ys = x_spots_um[s], y_spots_um[s]
        for i in range(ny):
            dy = y_grid_um[i] - ys
            dy_sq = dy * dy
            if dy_sq > cutoff_r_sq: continue
            for j in range(nx):
                dx = x_grid_um[j] - xs
                r_sq = dx * dx + dy_sq
                if r_sq < cutoff_r_sq:
                    current_d = updated_depth[i, j]
                    w_deeper_sq = w_sq * (1.0 + (current_d / zR_um)**2)
                    F0_deeper = F0 * (w_sq / w_deeper_sq)
                    att_factor = np.exp(-current_d / D_sat)
                    fluence = F0_deeper * np.exp(-2.0 * r_sq / w_deeper_sq) * att_factor
                    if fluence > effective_F_th:
                        updated_depth[i, j] += delta_um * np.log(fluence / effective_F_th)
    return updated_depth

class LaserSimApp(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Laser Ablation Simulator (macOS Desktop)")
        self.geometry("1200x800")

        # 控制面板
        control_frame = ttk.Frame(self)
        control_frame.pack(side=tk.LEFT, fill=tk.Y, padx=10, pady=10)

        ttk.Button(control_frame, text="🚀 開始模擬", command=self.run_simulation).pack(pady=10)

        # Matplotlib 畫布
        self.fig = plt.figure(figsize=(10, 8), dpi=100)
        self.canvas = FigureCanvasTkAgg(self.fig, master=self)
        self.canvas.get_tk_widget().pack(side=tk.RIGHT, fill=tk.BOTH, expand=True)

    def run_simulation(self):
        self.fig.clear()
        # 簡單計算展示
        x = np.linspace(-20, 20, 100)
        y = np.linspace(-10, 10, 100)
        X, Y = np.meshgrid(x, y)
        Z = np.exp(-(X**2 + Y**2)/50.0)

        ax1 = self.fig.add_subplot(2, 3, 1, projection='3d')
        ax1.plot_surface(X, Y, -Z, cmap='viridis')
        ax1.set_title("3D Surface")

        ax5 = self.fig.add_subplot(2, 3, 5)
        ax5.plot(y, np.mean(Z, axis=1), 'g-')
        ax5.set_title("Average Depth vs. Y-Position")
        ax5.invert_yaxis()

        self.fig.tight_layout()
        self.canvas.draw()

if __name__ == "__main__":
    app = LaserSimApp()
    app.mainloop()
```
